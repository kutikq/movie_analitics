```python
%timeit controvers = rat_movies.top_controversial(10)

controvers = rat_movies.top_controversial(10)

print("Топ-10 фильмов с наибольшим разбросом оценок:")
for title, dispers in controvers.items():
    print(f"Разброс оценок: {dispers:.2f} | {title}")
```

Такс-такс, получались крайне занимательные результаты:

показатель дисперсии у фильма "My Fair Lady" оказался равным 5.06, коазалось бы вышел за границы оценки и не может существовать, однако, согласно формуле:

Дисперсия ($\sigma^2$) вычисляется как средний квадрат отклонений от среднего арифметического:

$$\sigma^2 = \frac{\sum (x_i - \bar{x})^2}{n}$$

Где:
* $x_i$ — оценка конкретного пользователя.
* $\bar{x}$ — средний рейтинг фильма.
* $n$ — количество оценок.

**Пример на наших данных:**
Предположим, у фильма всего две оценки: **0.5** (ужасно) и **5.0** (шедевр).
1. Находим среднее ($\bar{x}$): $(0.5 + 5.0) / 2 = 2.75$.
2. Считаем отклонения для каждой оценки:
   * $(0.5 - 2.75)^2 = (-2.25)^2 = 5.0625$
   * $(5.0 - 2.75)^2 = (2.25)^2 = 5.0625$
3. Находим среднее этих квадратов:
   $$\sigma^2 = \frac{5.0625 + 5.0625}{2} = 5.06$$

**Вывод:** Значение **5.06** — это максимально возможный «разброс» для шкалы от 0.5 до 5. Это математическое доказательство того, что аудитория разделилась на два непримиримых лагеря, выходит фильм либо ненавидят либо обожают, удивительное явление!

*Также хочется отметить что среди противоречевых фильмов нашелся неожиданный гость - "Криминальное чтиво"*

**Глава 4: Лингвистический портрет кино (Анализ тегов)**

Если оценки — это «математика» предпочтений, то теги — это их «философия». С помощью класса Tags мы проанализируем, какие смыслы пользователи вкладывают в просмотренные фильмы.

Мы изучим не только популярные темы, но и структуру самих тегов: насколько многословны наши зрители и какие уникальные ключевые слова объединяют разные картины. Это позволит нам понять, является ли система тегирования просто набором меток или полноценным инструментом описания киноязыка.

Для начала определим какие самые популярные тэги у нас вообще присутсвуют

```python
%timeit popular_tags = tags.most_popular(10)

popular_tags = tags.most_popular(10)

print("топ-10 самых гоярчих тем:\n")

for tag, count in popular_tags.items():
    print(f"{tag:<25} | Упоминаний: {count}")
```

При анализе тегов невозможно игнорировать абсолютного лидера — «In Netflix queue». Для современного зрителя это может звучать странно, но за этими тремя словами скрывается целая эпоха кинематографа.

Как это работало в 90-х и 2000-х: В то время Netflix не был стриминговым сервисом. Это был сервис проката DVD по почте. Пользователи составляли список фильмов, которые хотели посмотреть (ту самую «очередь»), и компания присылала им физические диски в знаменитых красных конвертах.

Почему этот тег так важен для нашего исследования?

Этот тег — «ископаемое» из времен, когда фильм нужно было ждать несколько дней из почтового ящика, а не кликать по экрану.

Теги как органайзер: Мы видим, что пользователи использовали MovieLens не только для оценки просмотренного, но и как персональный менеджер задач. Люди помечали фильмы «в очередь», чтобы не забыть заказать их на дисках.

Поведенческий паттерн: Тот факт, что техническая пометка обходит по популярности даже «Sci-Fi» или «Disney», доказывает: в те годы возможность организовать свой досуг была для людей важнее, чем художественное описание жанров.

Ну и конечно упомянем что наиболее поппулярным описательным тегом являеется 'atmospheric', что подтверждает популярность таких жанров как Drama и Romantic(нашли в самом начале исследования)


Как мы увидели у нас используются теги как органайзер, а попробуем найти самые длинные тэги и оценим какого они характера

Мы разобрались с нашими данными - *"маленьким прудиком"*, пора перейти в большой океан -  *сайт "IMDB"*

Класс Links позволяет нам получить идентификаторы IMDb, которые открывают доступ к полным рецензиям, кассовым сборам и именам режисеров на крупнейшем кинопортале мира.


Позволим себе небольшое лирическое отступление:

Процесс получения данных с IMDB оказался нетривиальной задачей. При большом количестве последовательных запросов сервер начинает возвращать «пустые» ответы, защищаясь от ботов. Именно поэтому в наших отладочных сообщениях мы видели, что спаршены не все данные.

Чтобы продемонстрировать работу нашего алгоритма «вживую», мы выберем 10 фильмов из середины списка и отправим прямые запросы к страницам IMDB.

```python
obj = Links("links.csv", "movies.csv")

movie_list = list(obj.all_data_dict.keys())[450:460]

fields = ['Director', 'Runtime', 'Rating']

%timeit res = obj.get_imdb(movie_list, fields)
res = obj.get_imdb(movie_list, fields)
```

Несмотря на сложности с получением свежих данных в реальном времени, накопленный кэш позволяет нам провести глубокий анализ финансовой и творческой составляющей киноиндустрии. Мы объединим наши локальные знания о рейтингах с глобальными показателями бюджетов и сборов.

```python
%timeit profitable = obj.most_profitable(5)
profitable = obj.most_profitable(5)

%timeit cost_per_min = obj.top_cost_per_minute(5)
cost_per_min = obj.top_cost_per_minute(5)

%timeit ratings_comparison = obj.dual_raiting(5)
ratings_comparison = obj.dual_raiting(5)

print("Экономический глобус:")
print("-" * 50)
for title, profit in profitable.items():
    print(f"Прибыль: ${profit:,} | {title[:30]}...")

print("\nстоимость производства (per min):")
print("-" * 50)
for title, cost in cost_per_min.items():
    print(f"${cost:<10} / мин | {title[:30]}...")
```

Чтобы завершить наше исследование, мы задействуем весь арсенал методов класса Links. Мы изучим не только деньги, но и время: найдем самых выносливых режиссеров, выделим фильмы-гиганты по хронометражу и сравним, насколько оценки обычных зрителей с IMDb коррелируют с жестким мнением критиков из Metascore.

```python
%timeit top_dirs = obj.top_directors(10)
top_dirs = obj.top_directors(10)

%timeit long_movies = obj.longest(10)
long_movies = obj.longest(10)

%timeit dual_rev = obj.dual_raiting(10) 
dual_rev = obj.dual_raiting(10) 

print("Мастеры режиссуры (Топ по количеству работ):")
print("-" * 55)
for name, count in top_dirs.items():
    print(f"{name:<30} | Фильмов в базе: {count}")

print("\nГиганты хронометража (Самые длинные фильмы):")
print("-" * 55)
for title, runtime in long_movies.items():
    print(f"{title[:35]:<35} | {runtime} мин.")

print("\nБитва мнений (Зрители vs Критики):")
print("-" * 55)
print(f"{'Название фильма':<35} | {'IMDb':^5} | {'Meta':^5}")
for title, scores in dual_rev.items():
    # Защита от None при выводе
    imdb_s = scores[0] if scores[0] is not None else "N/A"
    meta_s = scores[1] if scores[1] is not None else "N/A"
    print(f"{title[:35]:<35} | {imdb_s:^5} | {meta_s:^5}")
```

Анализ производственных стратегий:

Эффективность инвестиций: Ранее мы видели, что фильм Waterworld лидирует по стоимости минуты проката ($1.29 млн/мин), однако в списке самых прибыльных его нет. Это подтверждает классический риск «перепроизводства» в Голливуде середины 90-х.

Диссонанс рейтингов: Сравнение через dual_raiting показывает интересную закономерность: фильмы с самым высоким баллом IMDb часто имеют более низкий Metascore. Это говорит о том, что массовый зритель более склонен прощать сюжетные огрехи ради эмоций, в то время как критики оценивают техническую безупречность.

Режиссерский фактор: Список топ-режиссеров позволяет увидеть тех, кто «держал» индустрию в те годы. Большое количество фильмов от одного автора в нашей выборке указывает на высокую степень доверия студий к определенным именам.

Хронометраж: Самые длинные фильмы (более 150-180 минут) коррелируют с высокими бюджетами, что логично: масштабные истории требуют масштабных затрат времени и денег.
